## Notebook Purpose:
This notebook processes videos saving a captioned video as well as transcripts of those captions. These captions have a sentence level and word level timestep which can be used later for summarization and detail capture

````python
import os
import sys
import time

sys.path.append('../')

from utils import VideoProcessingPipeline
````

````python
data_dir = '../data/videos'
video_example = os.path.join(data_dir, os.listdir(data_dir)[0])
````

````python
print('Video example:', video_example)
````

````python
pipeline = VideoProcessingPipeline(model_size="medium",
                                   extract_frames_interval=5,
                                   output_dir='../data/processed_videos')
````

````python
start_time = time.time()
videos = os.listdir(data_dir)
videos = [os.path.join(data_dir, video) for video in videos]
for i in range(0, len(videos), 5):
    print('Processing videos:', f"{i}:{i+5}")
    video_paths = videos[i:i+5]
    results = pipeline.batch_process(video_paths)
end_time = time.time()
print(f"Processing {len(videos)} videos took {end_time - start_time:.1f}s")
````
